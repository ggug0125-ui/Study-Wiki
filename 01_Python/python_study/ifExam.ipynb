```python
# if문은 조건문으로 판단하여 참(True) 거짓(False)로 핀딘하는 코드
# if 조건 :
#   조건이 참일 때 실항해는 문
#elif 조건 :
#   위 조건이 false일때 또 비교하는 조건문
#else :
#   위 조건이 모두 해당하지 않을때 실행하는 조건문
# 학점 계산용 코드

avg = int(input("평균을 입력하세요"))
if avg >= 90:
    grade = "A"  # 90점 이상이면 A
elif avg >= 80:
    grade = "B"
elif avg >= 70:
    grade = "C"
else:
    grade = "F"

    print(" 당신의 학점은 : " + grade + "입니다.")
```

```python
# while  무한반복 반복문 끝을 내가 낸다, for문은 리스트에 값을 다구함 종료 끝이 있다
나무찍어 = 0
while 나무찍어 < 10:
    나무찍어 = 나무찍어 + 1
    print("나무를 %d번 찍었습니다." % int(나무찍어))
    if 나무찍어 == 10:
        print("나무가 넘어갔어요~~~")
```

```python
coffee = 10
money = 300
while money:
    print("돈을 받았으니 커피를 줄께~")
    coffee = coffee -1
    print("남은 커피의 양은 %d 개 입니다." % coffee)
    if coffee == 0:
        print("커피가 다 떨어졌어요, 판매중지됩니다.")
        break
```

```python
coffee = 10   # 커피 10개있다
while True:
    money = int(input("돈을 넣어 주세요:"))  # 숫자로 계산해야해서 숫자변환해주기
    if money == 300:  #커피값은 300원이야
        print("커피를 줄께요")
        coffee = coffee -1
    elif money > 300: # 300원 보다 클경우는 잔도주고 커피줄께
        print("잔돈 %d를 주고 커피를 드릴께요" %(money -300))
        coffee = coffee -1
    else: # 300원보다 작을경우는 돈다시 주고 커피 안줘
        print("돈을 다시 돌려주고 커피를 주지 않을꺼야")
        print("남은 커피의 양은 %d개 입니다." %coffee)
    if coffee == 0:  #커피 0일때는 커피없어
        print("커피없어~ 판매 중지할꺼야")
        break
```

```python
marks = [90,25,67,45,80] #점수
number = 0 # 학생번호
for mark in marks: # 90 25 67 45 80 순서대로 mark 대입
    number = number + 1
    if mark >= 60:
        print("%d번 학생은 합격입니다" %number)
    else:
        print("%d번 학생은 불합격입니다" %number )
```

```python
mark = [90,25,67,45,80]
number = 0
for mark in mark:
    number = number + 1
    if mark < 60:  # 60점이하인 학생은 참이되어 continue 로 수행, 나머지는 프린트 수행됨
        continue
    print("%d번 학생 축하합니다. 합격입니다" %number)
```

```python
names = ['홍길동','김철수','이영희']
scores = [85,92,78]
for name, score in zip(names, scores):  #zip 두개이상의 리스트를 동시에 순회하고 싶을때
    print(f"{name}: {score}점")
```

```python
names = ['홍길동', '김철수', '이영희']
korean = [85, 92, 78]
english = [90, 88, 95]
for name, kor, eng in zip(names, korean, english):
    print(f"{name}: 국어 {kor}점, 영어 {eng}점")

```
