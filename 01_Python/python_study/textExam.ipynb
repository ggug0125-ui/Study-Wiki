```python
f = open("새파일.txt","w")
# open 파일을 열때 사용 (파일을 열땐 나만열아, 읽기모드로는 가능해)
# 옵션 : r -> read 읽기전용
#       W -> writer 쓰기가능
#       a -> append 추가가능(마지막행에 추가)
# 파일을 열때 경로를 지정하지 않으면 같은 위치에 파일을 찾거나 쓴다.
f.close()
# 파일은 윈도우 정책에서 단일처리를 한다.(열고 작업이 끝나면 닫고 나와야 함)
# 안닫으면 다른 프로그램에서 열수가 없다.
```

```python
f = open("c:/mbc/새파일.txt","w")
f.close()
#주의사항 경로에 / 를 붙이는게 일반적이고
# 만약 \를 사용하고 싶으면 \\두개를 써야한다  c:\\mbc\\새파일.txt
# \t 는 탭,  \n은 엔터  , \는 이스케이프 문자로 기능이 있는 문자열
```

```python
f = open("c:/mbc/새파일1.txt","w")
for i in range(1,11) :  # range (1~10까지 값을 가지고있다)
    data = "%d번째 줄입니다. \n" % i
    #      정수처리        엔터    for = i
    f.write(data) # .write()기록
    # 파일에 저장

f.close()
```

```python
# 파일있는 내용을 가져와 보자!!!
# .readline() 파일에 있는 내용 줄단위로 가져오기
f = open("c:/mbc/새파일1.txt","r")
line = f.readline() #첫줄 가져오기가 기본값이야~
print(line)
f.close()
```

```python
#모든줄을 읽어 화면에 출력하고 싶으면
f = open("c:/mbc/새파일1.txt","r")
while True: #False 처리될때까지 반복문
    line = f.readline()  # 1줄씩 읽는다.
    if not line :
        break    #line이 없으면 break로 중지
    else:
        print(line)

f.close()

#메모장 내용에 엔터가 적용이 되어 있는 상태임 \n,\r(타자기형식) 라인피드 캐리지 리턴
```

```python
# .readlines() 메모장 파일을 리스트로 한번에 불러온다.
# 파일의 모든줄을 읽어 리스트 형태로 반환
# 반환값은 문자열 리스트['첫번째줄\n,두번째줄\n'.......)
# 단점 : 파일 내용을 모두 메모리에 로드하기 때문에 파일크기가 클 경우 메모리부족
# 장점 : 파일의 모든 줄을 한번에 가져와 처리할때 편해

f = open("c:/mbc/새파일1.txt","r")
lines = f.readlines()
print(lines)
for line in lines:
    print(line)
f.close()
```

```python
# 메모장에 엔터효과인 \n 을 지우고 싶으면 strip을 써라~
# strip()은 문자열내에서 원하는 문자열 또는 공백을 제거함
# lstrip() 원하는 문자열 또는 공백(엔터)을 왼쪽에서 제거
# rstrip() 원하는 문자열 또는 공백(엔터)을 오른쪽에서 제거
f = open("c:/mbc/새파일1.txt","r")
lines = f.readlines()
for line in lines:
    line = line.strip()   #스트립은 줄바꿈을 지우는거다~!
    print(line)
    f.close()
```

```python
#read()를 이용하여 파일 문자열 읽기
#파일의 내용 전체를 문자열로 반환
f = open("c:/mbc/새파일1.txt","r")
data = f.read()  # 리드를 이용하면 전제문자열 가져와~ 줄바꿈 기호도 써버린거야~ 프린트한내용 고대로 가져오는거야
print(data) # for문을 사용하지않고 전제를 읽어 출력
f.close()
```

```python
#for를 이용해 줄단위로 가져오기
f = open("c:/mbc/새파일1.txt","r")
# 파일객체는 기본적으로 for문과 같이 함께 사용하여 줄단위로 읽기도 가능
for line in f:
    print(line)
f.close()
```

```python
# 파일의 새로운 내용을 추가 .append -> a 모드~~
# w로 파일을 열면 이미 존재하는 파일이면 그 내용이 사라지고 덮어쓰기가 된다~!!!!!!
# a모드는 원래 있던 값을 유지하면서 새로운 값을 맨 아래에 추가~~
f = open("c:/mbc/새파일1.txt","a")   # append모드 추가모드
f.write("---------------------")
for i in range(11,20):
    data = "%d번째 줄입니다.\n" %i
    f.write(data)
f.close()
```
