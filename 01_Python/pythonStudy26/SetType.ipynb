```python
# 집합자료형 (set)
# 로또 주머니와 같이 중복을 허용하지 않고 순서가 없는
# 데이터들의 모임이다.
# 교집합, 합집합, 차집합 등의 집합 연산을 쉽게 처리할수있다.

s1 = set([1,2,3])  #set() 값을 넣으면 된다.
print(s1)
```

```python
s2 = set("hello")   #중복된 값없어짐, 순서도 지맘대로
print(s2)
```
