```python
# 딕셔너리 타입은 key : value 로 이루어 진다.
# { key1 : value1, key2 : value2......}
# 대부분 웹 프로그랜 언어는 json이라는 값으로 처리 하는데
# 유사한 형태이다

dic = {"name" : "kkw" , "age" : 40 }
#        k        v       k      v


type(dic)
print(dic["name"])
#            k
```

```python
a = {1 : "hi"}
print(a[1])   # 인덱스 대신해 키를 이용해 가져온다
#       k
```

```python
b = {"a" : [1,2,3]}  # 키에다가 리스트값을 벨류로 넣을수있다
#     k       v
print(b["a"])
print(b["a"][1])
```

```python
print(dic)
```

```python
dic['email'] = "kkw@mbc.com"  # dic변수에 email key를 갖는 값을 추가하는 방법
print(dic)
```

```python
# dic 변수에 age를 삭제해보자  지울때는 del!!!
del dic["age"]
print(dic)
```

```python
# dic 변수에 name 를 변경할수있다
dic['name'] = "김기원"
print(dic)
```

```python
# key의 리스트를 생성하려면 : a.keys()   키만 뽑아오기
dic.keys()
```

```python
#키의 값들을 리스트로 생성하려면   벨류만 뽑아오기
dic.values()
```

```python
#key 와 value 의 쌍으로 튜플처리하려면     키 벨류 둘다 같이 뽑아오고 튜플처리됨
dic.items()
```

```python
# get을 이용해 키를 넣으면 값이 나오도록      키를 널고 값을 뽑아온다
dic.get("name")
```

```python
# key에 해당하는 값을 가져올때 없는 키를 넣으면 에러발생
dic["age"]  # 없는 키를 가져오려면 오류발생 key Error
# 이것은 프로그램 실행중 치명적인 오류로 강제 종료 된다.
```

```python
#get을 쓰면 없는 값을 찾아도 에러발생하지 않는다  키를찾을때 .get을 이용해서 찾는다
dic.get("age")
```

```python
dic.get("age","값없음")
#         k      y     k가 없을때 출력뮨
```

```python
print(dic)
```

```python
"name" in dic  # in 명령어는 리스트나,튜플, 딕셔너리 안에 값이 있는지 물어보는거임
```

```python
"김기원" in dic  # 키만본다 벨류입력하면 검색이 안된다~
```

```python
# key를 이용해서 value를 가져오는게 있다 - pop 을 이용 (값을 꺼내와 변수에 넣고 삭제한다)
print(dic)
```

```python
email = dic.pop("email")
print(email)
```
