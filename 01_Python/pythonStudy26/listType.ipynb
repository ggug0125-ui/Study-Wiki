```python
# 리스트에 C R U D 기법을 적용해보자
# CREATE -> 생성, 추가, 삽입  .append()
# READ -> 전체 리스트보기 또는 1개 자료보기 if문으로 in 돌리(indea 맞는지 틀린지)  .index()
# UPDATE -> 자료수정 passwords[3] = "4486"
# DELETE -> 자료삭제 del passwords[3] .remove(3)  .pop(3)
```

```python
# 리스트의 자료를 추가해보자
ids = ["kkw","lhj","ksb"]  # 회원 id
ids.append("MBC")    # 회원가입할때 마지막 빈칸에 계속 추가 입력됨 - 회원추가~!!!
print(ids)
```

```python
passwords = ["1234","4321","4567"]
passwords.append("8888")    #.append 는 추가다~!!!!!!!!
print(passwords)
```

```python
# 리스트에 있는 값 가져오기
ids.index("MBC")
```

```python
# MBC ID에 인덱스를 알아봐 패스워드 값을 찾아온다
passwords[3]
```

```python
# 알아온 password 값을 4486으로 수정해본다  수정은"="넣어
passwords[3] = "4486"
print(passwords)
```

```python
# 3번째 인덱스에 암호를 삭제해본다  ->메세지 없이 지워짐
passwords.remove("4486")
print(passwords)
```

```python
# pop은 삭제되는 내용을 출력하고 삭제한다.
# passwords.pop() -> 맨 마지막 자료를 삭제
# passwords.pop(2) -> 2 인덱스를 삭제한다
passwords.pop(2)   #  프린트를 안해도 알려준다 요것이 삭제된거다~!!
```
